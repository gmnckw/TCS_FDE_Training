# Classes and Objects

Bundle data and behaviour together with object-oriented programming (OOP).

**You will learn to:**

- Explain a class, an object, an attribute and a method
- Write a constructor with `__init__`
- Tell instance attributes from class attributes
- Make objects print clearly with `__str__` and `__repr__`
- Use `@dataclass` for simple data holders

## 1. Why classes?

A **class** is a blueprint. An **object** (or **instance**) is one thing built from that blueprint.
Think of `ChatMessage` as the blueprint and each individual message as an object.

| Term | Meaning |
|---|---|
| Class | The blueprint: `class Dog:` |
| Object / instance | One thing made from it: `Dog("Buddy")` |
| Attribute | Data stored on an object: `dog.name` |
| Method | A function that belongs to a class: `dog.bark()` |
| `self` | The object the method is working on |

## 2. Your first class

```python
class Dog:
    def __init__(self, name, age):          # the constructor
        self.name = name
        self.age = age

    def bark(self):
        return f"{self.name} says Woof!"


buddy = Dog("Buddy", 3)
lucy = Dog("Lucy", 5)

print(buddy.name, buddy.age)
print(buddy.bark())
print(lucy.bark())
print(type(buddy))
```

`__init__` runs automatically when you create an object. Each object keeps its **own** attribute values.

```python
buddy.age += 1
print(buddy.age, lucy.age)       # lucy is unchanged
print(buddy is lucy, buddy == Dog("Buddy", 4))   # different objects
```

## 3. Methods that change state

```python
class TokenBudget:
    def __init__(self, limit):
        self.limit = limit
        self.used = 0

    def spend(self, tokens):
        self.used += tokens

    def remaining(self):
        return self.limit - self.used

    def is_exhausted(self):
        return self.remaining() <= 0


budget = TokenBudget(1000)
budget.spend(400)
budget.spend(250)

print(budget.used, budget.remaining(), budget.is_exhausted())
```

## 4. Instance vs class attributes

```python
class Model:
    provider = "Anthropic"             # class attribute: shared by all objects
    count = 0

    def __init__(self, name):
        self.name = name               # instance attribute: one per object
        Model.count += 1


a = Model("haiku")
b = Model("sonnet")

print(a.provider, b.provider)
print(Model.count)
```

## 5. Readable objects: `__str__` and `__repr__`

- `__str__` is the friendly text used by `print`.
- `__repr__` is the unambiguous text used when debugging. Aim for something that looks like the code to create the object.

```python
class ChatMessage:
    def __init__(self, role, content):
        self.role = role
        self.content = content

    def __str__(self):
        return f"{self.role}: {self.content}"

    def __repr__(self):
        return f"ChatMessage(role={self.role!r}, content={self.content!r})"

    def to_dict(self):
        return {"role": self.role, "content": self.content}


message = ChatMessage("user", "Hello!")
print(message)
print(repr(message))
print([message])
print(message.to_dict())
```

## 6. Objects that hold other objects

```python
class Conversation:
    def __init__(self, system_prompt=""):
        self.system_prompt = system_prompt
        self.messages = []

    def add(self, role, content):
        self.messages.append(ChatMessage(role, content))

    def to_api_format(self):
        return [m.to_dict() for m in self.messages]

    def __len__(self):
        return len(self.messages)


chat = Conversation("You are concise.")
chat.add("user", "What is OOP?")
chat.add("assistant", "Objects that bundle data and behaviour.")

print(len(chat))
print(chat.to_api_format())
```

## 7. `@dataclass` for simple data holders

When a class mostly stores data, `@dataclass` writes `__init__`, `__repr__` and `==` for you.

```python
from dataclasses import dataclass, field


@dataclass
class Document:
    title: str
    pages: int = 1
    tags: list = field(default_factory=list)      # a safe default list


doc = Document("Handbook", 12)
doc.tags.append("hr")

print(doc)
print(doc == Document("Handbook", 12, ["hr"]))
```
